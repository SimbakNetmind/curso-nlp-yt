# 🎯 **Stemming**

Stemming es una técnica de procesamiento de lenguaje natural (PLN) que reduce las palabras a su forma raíz o 'stem', eliminando prefijos y sufijos. El objetivo es tratar palabras morfológicamente similares como la misma palabra (ej., 'corriendo', 'corredor', 'corre' se reducirían a 'corr'). Esto ayuda a reducir la dimensionalidad del vocabulario y puede mejorar el rendimiento de los modelos.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer

nltk.download("stopwords")
nltk.download("punkt") # Changed from 'punkt_tab' for standard punkt tokenizer

stemmer = SnowballStemmer("spanish")

def tokenize_and_stem(text):
    tokens = word_tokenize(text.lower())
    stems = [stemmer.stem(token) for token in tokens if token.isalpha()]
    return ' '.join(stems)

df['news_stemmer'] = df['news'].apply(tokenize_and_stem)

display(df['news_stemmer'].head()) # Displaying head for brevity and clarity
```

```python
print(df['news_lemma'][3])
```

### 👉 **Nuevo Modelo con Texto Stemmed**

En este bloque, se entrena y evalúa nuevamente el modelo de clasificación, pero esta vez utilizando la columna de noticias a la que se le aplicó stemming (`news_stemmer`). Se espera que, al reducir las palabras a sus raíces, el modelo pueda generalizar mejor y potencialmente mejorar su precisión.

```python
# Separamos los datos en variables de entrada y etiquetas utilizando el texto 'stemmed'
X = df['news_stemmer']
y = df['Type']

# Dividir los datos en conjuntos de entrenamiento y prueba
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Vectorizar los nuevos datos 'stemmed'
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
X_train_transformed = vectorizer.fit_transform(X_train)
X_test_transformed = vectorizer.transform(X_test)

# Entrenar un nuevo modelo MultinomialNB
from sklearn.naive_bayes import MultinomialNB
from sklearn import metrics
model = MultinomialNB()
model.fit(X_train_transformed, y_train)

# Medir el rendimiento del modelo
y_pred = model.predict(X_test_transformed)

print(f"Precisión del modelo con texto stemmed: {metrics.accuracy_score(y_test, y_pred)}")
```

### 🎯 **Lematización**

La lematización es un proceso más avanzado que el stemming. Su objetivo es reducir las palabras a su forma base o 'lema', que es una palabra real que se encuentra en un diccionario (ej., 'corriendo', 'corredor', 'corre' se reducirían a 'correr'). Esto implica un análisis morfológico y contextual, lo que a menudo resulta en una representación del texto más precisa y significativa para los modelos.

```python
!pip install spacy -q
!python -m spacy download es_core_news_sm -q
```

```python
import spacy

# Cargar el modelo en español
nlp = spacy.load('es_core_news_sm')

def lemmatize_text(text):
    # Procesar el texto convirtiéndolo a minúsculas
    doc = nlp(text.lower())
    # Extraer el lema de cada token si es una palabra alfabética
    lemmas = [token.lemma_ for token in doc if token.is_alpha]
    # Unir los lemas en una sola cadena de texto
    return ' '.join(lemmas)

# Aplicar la función de lematización a la columna 'news' y guardar en 'news_lemma'
df['news_lemma'] = df['news'].apply(lemmatize_text)

# Imprimir las primeras filas de la nueva columna 'news_lemma'
display(df['news_lemma'].head())
```

### 👉 **Datos de entrenamiento transformados**

Este fragmento muestra la representación en formato de matriz dispersa de los datos de entrenamiento vectorizados (`X_train_transformed`). Indica el tipo de datos, el número de elementos almacenados y la forma (número de documentos, tamaño del vocabulario). Esta matriz es la entrada numérica para el modelo de clasificación.

```python
# Separamos los datos en variables de entrada (X) y etiquetas (y)
# Nota: Puedes usar 'news_stemmer' o 'news_lemma' según tu columna disponible
X = df['news_lemma']
y = df['Type']

# Dividir los datos: 80% entrenamiento y 20% prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

# Convertir el texto a vectores numéricos (Bolsa de Palabras)
vectorizer = CountVectorizer()
X_train_transformed = vectorizer.fit_transform(X_train)
X_test_transformed = vectorizer.transform(X_test)

# Definir y entrenar el modelo Multinomial Naive Bayes
model = MultinomialNB()
model.fit(X_train_transformed, y_train)

# Realizar predicciones con los datos de prueba
y_pred = model.predict(X_test_transformed)

# Imprimir la precisión del modelo
print(metrics.accuracy_score(y_test, y_pred))
# Salida esperada aproximada: 0.8155...
```
