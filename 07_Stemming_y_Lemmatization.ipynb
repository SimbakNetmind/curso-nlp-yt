### 👉 ***Stemming con NLTK (SnowballStemmer)***Este bloque de código utiliza la librería `NLTK` para realizar el *stemming* en español. El `SnowballStemmer` es un algoritmo de stemming que elimina prefijos y sufijos de las palabras para encontrar su raíz (stem), la cual no siempre es una palabra real en el idioma.

```python
from nltk.stem import SnowballStemmer

#Crear el steammer en español
stemmer = SnowballStemmer('spanish')

#Probarlo en la palabra "caminar"
print(stemmer.stem('caminando'))
print(stemmer.stem('caminar'))
print(stemmer.stem('caminó'))
```

### 👉 ***Lemmatización con SpaCy***Esta sección se enfoca en la importación y uso de la librería `SpaCy` para realizar la *lemmatización*. A diferencia del stemming, la lemmatización reduce las palabras a su forma base léxica o de diccionario (lema), lo que garantiza que la palabra resultante sea una palabra real y válida en el idioma.

### 👉 ***Aplicación de Lemmatización con SpaCy***En este bloque, se carga el modelo de spaCy para español, se procesa un texto de ejemplo ('caminar caminando caminar') y luego se itera sobre cada token para imprimir la palabra original (`token.text`) y su forma lematizada (`token.lemma_`). Esto demuestra cómo spaCy puede identificar la forma base léxica de las palabras, incluso si cambian su terminación.

```python
import  spacy

# Cargar el modelo en español
nlp = spacy.load('es_core_news_sm')

# Crear un documento
doc = nlp('caminar caminando caminó')

# Imprimir  el texto y el leman de cada token

for token in doc:
    print(token.text,'->', token.lemma_)
```
