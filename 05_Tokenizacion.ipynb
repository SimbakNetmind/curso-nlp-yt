# ***Tokenización***


## **¿Qué es la Tokenización en NLP?**

La **tokenización** es el proceso de dividir una secuencia de texto en unidades más pequeñas llamadas *tokens*. Estos tokens pueden ser palabras, caracteres o subpalabras. El objetivo principal de la tokenización es preparar el texto para su posterior análisis, ya que la mayoría de los modelos de NLP trabajan con estas unidades discretas en lugar de con el texto completo.

### **¿Por qué es importante la Tokenización?**

1.  **Análisis de Texto**: Permite a los algoritmos entender y procesar el texto de manera estructurada.
2.  **Reducción de Complejidad**: Simplifica el texto en componentes manejables, facilitando tareas como la búsqueda, el análisis de sentimientos o la traducción automática.
3.  **Extracción de Características**: Los tokens son a menudo las características de entrada para modelos de aprendizaje automático.

### **Tipos Comunes de Tokenización:**

*   **Tokenización por Palabras**: Divide el texto en palabras individuales. Es el método más común. Los ejemplos en este cuaderno (`texto.split()`) son una forma simple de tokenización por palabras.
*   **Tokenización por Oraciones**: Divide el texto en oraciones.
*   **Tokenización por Subpalabras (Subword Tokenization)**: Divide las palabras en unidades más pequeñas, lo cual es útil para manejar palabras desconocidas (Out-Of-Vocabulary, OOV) y reducir el tamaño del vocabulario, especialmente en idiomas con morfología rica.



---
## 🎯 **Ejemplo de tokenización**

El código siguiente (texto ="Hola, ¿Cómo estás?")

tokens =texto.split()

print(tokens)

toma una cadena de texto y la divide en una lista de palabras (tokens) utilizando el método split(). Este es un método básico de tokenización por espacios en blanco.

```python
texto ="Hola, ¿Cómo estás?"
tokens =texto.split()
print(tokens)
```

### **Hacer las palabras a mínusculas**

El código subsiguiente texto = "Hola, ¿Cómo estás?"

texto = texto.lower()

print(texto)

 Primero convierte toda la cadena de texto a minúsculas (texto.lower()) para asegurar uniformidad, lo cual es útil antes de la tokenización para tratar palabras como 'Hola' y 'hola' como el mismo token. Luego, realiza la tokenización nuevamente con split().

```python
texto = "Hola, ¿Cómo estás?"
texto = texto.lower()
print(texto)

tokens = texto.split()
print(tokens)

```
