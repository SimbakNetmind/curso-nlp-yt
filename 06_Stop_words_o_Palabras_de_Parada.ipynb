#  📌 ***Stop Words***

## **¿Qué son las Stop Words?**

Las **Stop Words** (o palabras vacías) son palabras muy comunes en un idioma (como "el", "la", "un", "una", "y", "o", "en", "de", etc.) que generalmente no aportan un significado sustancial al contenido semántico de un texto. En el procesamiento del lenguaje natural (NLP), estas palabras son a menudo filtradas o eliminadas durante las fases de preprocesamiento del texto.

### **¿Por qué se eliminan las Stop Words?**

1.  **Reducción de Dimensionalidad**: Eliminar las *stop words* reduce el número de términos únicos en el vocabulario de un corpus de texto, lo que a su vez disminuye la dimensionalidad de los datos. Esto puede hacer que los modelos de machine learning sean más eficientes y requieran menos recursos computacionales.
2.  **Mejora de la Eficacia**: Al eliminar palabras que no contribuyen mucho al significado, los algoritmos pueden centrarse en los términos más relevantes y significativos. Esto es particularmente útil en tareas como la clasificación de texto, la recuperación de información o el análisis de sentimientos.
3.  **Reducción de Ruido**: Las *stop words* pueden introducir "ruido" en el análisis. Al eliminarlas, se mejora la calidad de los datos para el procesamiento posterior.

### **Consideraciones:**

*   La lista de *stop words* puede variar dependiendo del idioma y del caso de uso. NLTK, por ejemplo, proporciona listas predefinidas para varios idiomas.
*   En algunos contextos específicos (por ejemplo, análisis de sintaxis o reconocimiento de entidades nombradas), las *stop words* podrían ser relevantes y no deberían eliminarse.

###  👉 ***Conjunto Set (Stop words)***

Crea un conjunto (set) de palabras vacías en español. Las palabras vacías son términos comunes (como 'el', 'la', 'un', 'una') que a menudo se eliminan del texto durante el procesamiento porque no añaden mucho significado al análisis.

```python
stop_words = set(stopwords.words('spanish'))

print(stop_words)
```

### 👉 ***Tokeniza el texto (word_tokenize)***

 tokeniza el texto definido previamente ("El gato es negro y el perro es blanco")

 usando la función word_tokenize de NLTK,

 que es más inteligente que split() al manejar puntuación y otros caracteres.

```python
tokens = word_tokenize(texto)
print(tokens)
```

### 👉 ***Filtrar palabras que no sean (Stop Words)***

(texto_filtrado = [word for word in tokens if not word in stop_words]

print(texto_filtrado))

filtra la lista de tokens obtenida en el paso anterior.

Recorre cada palabra y la incluye en texto_filtrado solo si no está presente en el conjunto de stop_words.

```python
texto_filtrado = [word for word in tokens if not word in stop_words]

print(texto_filtrado)
```

## **Hacer todo el texto en mínusculas para elminar (El)**

código (texto = texto.lower())

tokens = word_tokenize(texto)

texto_filtrado = [word for word in tokens if not word in stop_words]

print(texto_filtrado)

vuelve a procesar el texto original. Primero, lo convierte a minúsculas (texto.lower()), luego lo tokeniza, y finalmente filtra las palabras vacías. Esto es para demostrar que al hacer el texto en minúsculas antes del filtrado, la palabra 'El' (que es una stop word en español) también es eliminada, cosa que no ocurrió en el paso anterior porque 'El' con mayúscula inicial no coincidía con 'el' en minúscula del conjunto de stop words.

```python
texto = texto.lower()
tokens = word_tokenize(texto)
texto_filtrado = [word for word in tokens if not word in stop_words]
print(texto_filtrado)
```
